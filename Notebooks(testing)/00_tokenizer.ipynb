```python
from transformers import GPT2Tokenizer

toknizer = GPT2Tokenizer.from_pretrained("gpt2")
```

```python
text = "Your cat is a lovely cat"
tokens = toknizer.tokenize(text)
```

```python
print(tokens)
```

```python
token_ids = toknizer.encode(text)
print(token_ids)
```
